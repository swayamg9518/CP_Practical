```python
# elif else statement for marks of students
# 90 and above - A
# 85-89 - B
# below 85 - C
marks = int(input("Enter your marks: "))
if marks >= 90:
    print("Grade: A")
elif marks >= 85:
    print("Grade: B")
else:
    print("Grade: C")   
    
```
